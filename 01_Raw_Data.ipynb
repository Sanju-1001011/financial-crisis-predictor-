# Stage 1: 
# Raw Data Ingestion & Cleaning
# This notebook loads the foundational historical macroeconomic dataset and removes any records where the actual crisis outcome is unknown. the code for this stage of the pipeline here.*

```python
import pandas as pd
import os

# Create folders if they don't exist
os.makedirs('data', exist_ok=True)
os.makedirs('models', exist_ok=True)

# 1. Load Raw Data
print("Loading raw data...")
df = pd.read_csv('01_Historical_Financial_Crisis_Macroeconomic_Data.csv')

# 2. Clean Target Variable
# We cannot train the AI on years where we don't know if a crisis happened
df = df.dropna(subset=['crisisJST'])

# 3. Save interim data for the next notebook
df.to_csv('data/01_cleaned_raw.csv', index=False)
print(f"Raw data cleaned. Shape: {df.shape}")
df.head()
```
